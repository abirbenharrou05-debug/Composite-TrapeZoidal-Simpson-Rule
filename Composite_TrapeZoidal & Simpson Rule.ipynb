```python
# Composite Trape Zoidal

def Composite_TrapeZoidal(function,a, b, n ):
    h = (b-a)/n
    S = function(a)+function(b)
    for i in range(1,n):
        S += 2 * function(a+i*h)
    return (h/2) * S 
```

```python
# Composite_Simpson_Rule

def Composite_Simpson_Rule(function,a,b,n):
    h=(b-a)/n
    I = function(a)
    for i in range(1,n):
        if (i % 2 != 0):
            I += 4* function(a+i*h)
        else : 
            I += 2 * function(a+i*h)
    I += function(b)
    I *= (h/3)
    return I
```

```python
import numpy as np

# Cube
def cube(x):
    return np.power(x,3)
```

```python
from scipy.integrate import quad

valeur_ref, erreur_estimee = quad(cube, 1, 2)


print("L'integrale exacte : ", valeur_ref)
print("Composite trapezoidal rule : ", Composite_TrapeZoidal(cube,1,2,4))
print("Composite Simpson Rule : ", Composite_Simpson_Rule(cube,1,2,4))


err_trap = abs(Composite_TrapeZoidal(cube, 1, 2, 4) - valeur_ref)
err_simp = abs(Composite_Simpson_Rule(cube, 1, 2, 4) - valeur_ref)
print("erreur trapèzes :", err_trap)   
print("erreur Simpson  :", err_simp)   
```
